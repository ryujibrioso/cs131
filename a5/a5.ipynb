```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Link to dataset: https://www.kaggle.com/datasets/waqi786/cats-dataset?resource=download

```python
# Load dataset
cats_df = pd.read_csv('C:/Users/ryuji/Downloads/archive/cats_dataset.csv')

# Display metadata
print(cats_df.info())
print(cats_df.describe())
```

```python
# Plot style
sns.set(style="whitegrid")

# 1. Box Plot: Weight distribution by breed
plt.figure(figsize=(12, 8))
sns.boxplot(x='Weight (kg)', y='Breed', data=cats_df, palette='Set2')
plt.title('Weight Distribution by Cat Breed')
plt.xlabel('Weight (kg)')
plt.ylabel('Breed')
plt.show()
```

```python
# 2. Scatter Plot: Age vs. Weight, colored by Gender
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Age (Years)', y='Weight (kg)', hue='Gender', data=cats_df, palette='coolwarm')
plt.title('Age vs. Weight of Cats')
plt.xlabel('Age (Years)')
plt.ylabel('Weight (kg)')
plt.show()
```

```python
# Increase figure size
plt.figure(figsize=(25, 10))

# Create the box plot with swarm plot overlay
sns.boxplot(x='Breed', y='Weight (kg)', hue='Gender', data=cats_df, palette='Set2')

# Swarm plot with jitter to spread out the points
sns.swarmplot(x='Breed', y='Weight (kg)', hue='Gender', data=cats_df, dodge=True, palette='Set2', size=5, alpha=0.7)

# Adjust the spacing between the x-axis labels
plt.xticks(rotation=45, ha='right', fontsize=12)

# Adjust the padding between x-ticks
plt.gca().margins(x=0.02)

# Titles and labels
plt.title('Weight Distribution by Cat Breed and Gender', fontsize=20)
plt.xlabel('Breed', fontsize=16)
plt.ylabel('Weight (kg)', fontsize=16)

handles, labels = plt.gca().get_legend_handles_labels()
plt.legend(handles[0:2], labels[0:2], title="Gender", bbox_to_anchor=(1.05, 1), loc='upper left')

# Adjust subplot parameters for more spacing around the plot
plt.subplots_adjust(left=0.05, right=0.9, top=0.9, bottom=0.2)


plt.show()
```
